```python
with conn.cursor() as cur:
    cur.execute("""
        SELECT 'Bronze' AS layer, COUNT(*) AS total_records
        FROM bronze.sales_raw

        UNION ALL

        SELECT 'Silver', COUNT(*)
        FROM silver.sales

        UNION ALL

        SELECT 'Gold', COUNT(*)
        FROM gold.sales_summary;
    """)

    for row in cur.fetchall():
        print(f"{row[0]}: {row[1]}")
```
